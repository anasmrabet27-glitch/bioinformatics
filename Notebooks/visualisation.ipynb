```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_excel(r"C:\Users\anasm\Downloads\test4.xlsx")

x = np.arange(len(df))
largeur = 0.4

plt.figure(figsize=(14, 6))

plt.bar(x - largeur/2, df["cdna_GC%"],
        largeur, label="cDNA")

plt.bar(x + largeur/2, df["cds_GC%"],
        largeur, label="CDS")

plt.xlabel("Transcrits")
plt.ylabel("GC (%)")
plt.title("Comparaison du GC % des cDNA et des CDS")
plt.xticks(x, df["id"], rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```
