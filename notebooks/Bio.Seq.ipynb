```python
from Bio.Seq import Seq

ADN = Seq("ATCGTGCGTCC")  
print("=== ANALYSE DE SÉQUENCE ADN ===")
print(f"Séquence: {ADN}")
print("ARN:", ADN.transcribe())
print("Protéine:", ADN.translate())
print("Complément:", ADN.complement())
print("Reverse complément:", ADN.reverse_complement())
    
    # Analyse des nucléotides
longueur = len(ADN)
A, T, G, C = ADN.count("A"), ADN.count("T"), ADN.count("G"), ADN.count("C")
    
print(f"\n=== COMPOSITION ===")
print(f"Longueur: {longueur}")
print(f"A: {A} ({A/longueur*100:.2f}%)")
print(f"T: {T} ({T/longueur*100:.2f}%)")
print(f"G: {G} ({G/longueur*100:.2f}%)")
print(f"C: {C} ({C/longueur*100:.2f}%)")
    
    # Contenu GC/AT
GC, AT = G + C, A + T
print(f"\n=== RATIOS ===")
print(f"GC: {GC} ({GC/longueur*100:.2f}%)")
print(f"AT: {AT} ({AT/longueur*100:.2f}%)")

# Exemple d'utilisation :
# analyser_adn("ATCGATCGATCG")
```
