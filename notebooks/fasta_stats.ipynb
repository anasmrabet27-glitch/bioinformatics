```python
from Bio import SeqIO

filename = r"C:....\rna.fna"

results = []

for record in SeqIO.parse(filename, "fasta"):
    seq = record.seq
    results.append({
        "ID": record.id,
        "Length": len(seq),
        "A": seq.count("A"),
        "T": seq.count("T"),
        "G": seq.count("G"),
        "C": seq.count("C"),
        "N": seq.count("N"),
    })
```

```python
import pandas as pd 
df = pd.DataFrame(results)
df
```

```python
df["GC%"]= (df["G"] + df["C"]) / df["Length"] * 100
df
```

```python
df.to_csv(
    r"C:....\BRCA1_transcri_analysis.csv"
)
```
