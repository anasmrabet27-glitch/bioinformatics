```python
from Bio import SeqIO
record = SeqIO.read(r"fasta_sequence","fasta")
print("ID:", record.id)
print("Description:", record.description)
print("longueur:", len(record.seq))
```

### Key concepts

- `SeqIO.read()` reads one sequence record.
- `record.id` gives the sequence identifier.
- `record.description` gives the FASTA description.
- `len(record.seq)` returns the sequence length.

```python
from Bio import SeqIO
for record in SeqIO.parse(r"fasta_sequence","fasta"):
    print("ID:", record.id)
    print("Description:", record.description)
    print("longueur:", len(record.seq))
    
```

### Reading multiple FASTA sequences

`SeqIO.parse()` reads a FASTA file and processes each sequence one by one.
* `record.id` → sequence ID
* `record.description` → sequence description
* `record.seq` → sequence
* `len(record.seq)` → sequence length
