```python
import gzip
from contextlib import contextmanager
from Bio import SeqIO

@contextmanager
def open_fastq(filename):

    if filename.endswith(".gz"):
        handle = gzip.open(filename, "rt")
    else:
        handle = open(filename, "r")

    try:
        yield SeqIO.parse(handle, "fastq")
    finally:
        handle.close()
```

```python
filename = path.file

total_reads = 0
reads_with_N = 0
total_bases = 0
total_N = 0
total_G = 0
total_C = 0
total_quality = 0
total_quality_bases = 0

with open_fastq(filename) as records:

    for record in records:

        seq = record.seq

        total_reads += 1
        total_bases += len(seq)

        total_N += seq.count("N")
        total_G += seq.count("G")
        total_C += seq.count("C")

        if "N" in seq:
            reads_with_N += 1

        qualities = record.letter_annotations["phred_quality"]
        total_quality += sum(qualities)
        total_quality_bases += len(qualities)
```

```python
if total-reads >0:
    reads_N_percent = (reads_with_N / total_reads) * 100
else:
    reads_N_percent = 0
if total_bases >0:
    N_percent = (total_N / total_bases) * 100
    GC_percent = ((total_G + total_C) / total_bases) * 100
else:
    N_percent = 0
    GC_percent = 0
mean_quality = total_quality / total_quality_bases
```

```python
print("===== FASTQ STATISTICS =====")

print("Total reads :", total_reads)
print("Total bases :", total_bases)

print("Reads containing N :", reads_with_N)
print("Reads with N (%) :", reads_N_percent)

print("Total N :", total_N)
print("N bases (%) :", N_percent)

print("GC (%) :", GC_percent)

print("Mean Phred quality :", mean_quality)
```

### FASTQ Statistics

This code opens a FASTQ or FASTQ.GZ file and calculates basic statistics, including:

* Total number of reads
* Total number of bases
* Reads containing `N`
* Percentage of reads containing `N`
* Total number of `N` bases
* `N` percentage
* GC percentage
* Mean Phred quality

The code processes the reads one by one and uses `if` conditions to avoid division by zero when the file is empty.
